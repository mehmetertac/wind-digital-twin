# Thermal ODE surrogate demo

Timeboxed talking point: the lumped gearbox thermal ODE is the **slow simulator**; a small MLP learns temperature trajectories from parameter sweeps × load profiles. Requires `pip install -e ".[pinn]"`. Logic lives in `wind_digital_twin.surrogate.thermal_surrogate`.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from wind_digital_twin.surrogate.thermal_surrogate import (
    benchmark_surrogate_speed,
    build_surrogate_dataset,
    train_surrogate,
)

ds = build_surrogate_dataset(horizon=144, n_profiles=8, n_params=24, seed=0)
model, x_mean, x_std, val_rmse = train_surrogate(ds, epochs=35)
bench = benchmark_surrogate_speed(ds, model, x_mean, x_std)
print(f"Holdout RMSE (°C): {val_rmse:.3f}")
print(f"ODE s/trajectory: {bench['ode_seconds_per_trajectory']:.5f}")
print(f"NN s/trajectory: {bench['nn_seconds_per_trajectory']:.7f}")
print(f"Speedup: {bench['speedup_factor']:.0f}x")
```
